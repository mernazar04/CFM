```python

import numpy as np
import matplotlib.pyplot as plt


# Загружаем реальные тестовые данные
ace_real = np.load("data/normalized/ace_test.npy")
wg_real = np.load("data/normalized/wg_test.npy")

# Загружаем сгенерированные данные
ace_generated = np.load("generated_ace.npy")
wg_generated = np.load("generated_wg.npy")


def plot_comparison(real, generated, class_name):
    frames = [0, 5, 10, 15]

    fig, axes = plt.subplots(2, 4, figsize=(14, 7))

    for j, frame in enumerate(frames):

        # Реальная траектория
        axes[0, j].imshow(
            real[0, frame],
            cmap="RdBu_r"
        )
        axes[0, j].set_title(f"Real, t={frame}")

        # Сгенерированная траектория
        axes[1, j].imshow(
            generated[0, frame],
            cmap="RdBu_r"
        )
        axes[1, j].set_title(f"Generated, t={frame}")

        axes[0, j].axis("off")
        axes[1, j].axis("off")

    fig.suptitle(f"{class_name}: Real vs Generated")
    plt.tight_layout()
    plt.savefig(f"{class_name.lower()}_comparison.png", dpi=150)
    plt.show()


plot_comparison(ace_real, ace_generated, "ACE")
plot_comparison(wg_real, wg_generated, "Wave-Gauss")

print("Visualization completed!")
```
